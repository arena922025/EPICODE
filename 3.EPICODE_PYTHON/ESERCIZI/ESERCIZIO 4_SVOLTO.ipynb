```python
#quanti prodotti pesano piu di 100
import pandas as pd
query="SELECT*FROM dimproduct"
df = pd.read_sql(query, engine)
quanti= (df["Weight"]>100).sum()
print("quanti prodotti hanno un valore superiore a 100:",quanti)
#quali invece:
prodotti_pesanti=df[df["Weight"]>100]
print("i prodotti con peso > 100 sono:",prodotti_pesanti[["ProductKey","Weight"]])
```

```python
#quanto costano in media i prodotti colonna:DealerPrice
import pandas as pd
query="SELECT*FROM dimproduct"
df = pd.read_sql(query, engine)
print("la media del prezzo dei prodotti è:",df["DealerPrice"].mean())

#valore arrotondato:
print("la media del prezzo dei prodotti è:",round(df["DealerPrice"].mean(),2))
```

```python
#quali sono i quartili dei prezzi
import pandas as pd
query="SELECT*FROM dimproduct"
df = pd.read_sql(query, engine)
print("i quartili sono:")
q1=df["DealerPrice"].quantile(0.25)
q2=df["DealerPrice"].quantile(0.50)
q3=df["DealerPrice"].quantile(0.75)
print("Q1:",q1,"Q2:",q2,"Q3:",q3)
```

```python
#ES.3 #qual'è il prezzo medio per i prodotti di colore blu
#per il colore rosso o nero
#prezzo massimo per i prodotti di taglia 42 e peso oltre i 10kg
# qual'è il nome inglese e il costo di produzione(StandardCost) di tutti i prodotti taglia 42,peso oltre 10kg e colore argento?
#visualizza StandardCost e il DealerPrice degli ultimi 20 elementi:identifica differenze,c'èun pattern,c'è qualche elemento che non lo segue?

#qual'è il prezzo medio per i prodotti di colore blu
query="SELECT*FROM dimproduct"
df = pd.read_sql(query, engine)
prodotti_blu=df[df["Color"]=="Blue"]
prezzo_medio_blu= prodotti_blu["DealerPrice"].mean()
print("la media del prezzo dei prodotti di colore blue è:",prezzo_medio_blu)

#arrotondato
query="SELECT*FROM dimproduct"
df = pd.read_sql(query, engine)
prodotti_blu=df[df["Color"]=="Blue"]
prezzo_medio_blu= prodotti_blu["DealerPrice"].mean()
print("la media arrotondata del prezzo dei prodotti di colore blue è:",round(prezzo_medio_blu,2))
```

```python
#per il colore rosso o nero
query="SELECT*FROM dimproduct"
df = pd.read_sql(query, engine)
prodotti_blu=df[df["Color"]=="Black"]
prezzo_medio_blu= prodotti_blu["DealerPrice"].mean()
print("la media del prezzo dei prodotti di colore nero è:",prezzo_medio_blu)

#arrotondato
query="SELECT*FROM dimproduct"
df = pd.read_sql(query, engine)
prodotti_blu=df[df["Color"]=="Red"]
prezzo_medio_blu= prodotti_blu["DealerPrice"].mean()
print("la media arrotondata del prezzo dei prodotti di colore rosso è:",round(prezzo_medio_blu,2))
```

```python
#prezzo massimo per i prodotti di taglia 42 e peso oltre i 10kg
query="SELECT*FROM dimproduct"
df = pd.read_sql(query, engine)
taglia42_e_peso_maggiore_di_10kg=(df["Size"] == "42") & (df["Weight"] > 10)
prezzo_massimo=df[taglia42_e_peso_maggiore_di_10kg]["DealerPrice"].max()
print("Prezzo massimo per i prodotti di taglia 42 e peso > 10kg:", round(prezzo_massimo))
```

```python
#qual'è il nome inglese e il costo di produzione(StandardCost) di tutti i prodotti taglia 42,peso oltre 10kg e colore argento?
query="SELECT*FROM dimproduct"
df = pd.read_sql(query, engine)
taglia42_e_peso_maggiore_di_10kg=(df["Size"] == "42") & (df["Weight"] > 10) & (df["Color"]=="Silver")
nome_inglese_e_costo_produzione= df.loc[taglia42_e_peso_maggiore_di_10kg,["EnglishProductName","StandardCost"]]
print("tutti i prodotti taglia 42,peso oltre 10kg e colore argento sono:\n",nome_inglese_e_costo_produzione)
```

```python
#ES.5 
#importa il dataset diabetes.csv
#dimensioni 
import pandas as pd
df=pd.read_csv("C:\\Users\\gabri\\Desktop\\PERCORSO EPICODE\\M3_PYTON\\PRATICA\\datasets\\beginner_datasets\\diabetes.csv")
#metadati e colonne
print("i metadati sono:")

print("Colonne:\n")
print(df.columns)

print("Tipi di dati:\n")
print(df.dtypes)

print("Info del DataFrame:\n")
df.info()

# anteprima di 5 righe
prime_5_righe=df.head()
print("Le prime 5 righe sono:\n",prime_5_righe)

#stampa descrittori statistici del dataset
descrittori_statistici=df.describe()
print("I descrittori_statistici sono:\n",descrittori_statistici)


#dati relativi a fasce d'età:20-29,30-39,40-50;
# Persone tra 20 e 29 anni
fascia_20_29 = df[(df["Age (years)"] >= 20) & (df["Age (years)"] <= 29)]

# Persone tra 30 e 39 anni
fascia_30_39 = df[(df["Age (years)"] >= 30) & (df["Age (years)"] <= 39)]

# Persone tra 40 e 50 anni
fascia_40_50 = df[(df["Age (years)"] >= 40) & (df["Age (years)"] <= 50)]

# Mostra quante righe ci sono in ciascuna fascia
print("Numero di persone tra 20-29 anni:", fascia_20_29.shape[0])
print("Numero di persone tra 30-39 anni:", fascia_30_39.shape[0])
print("Numero di persone tra 40-50 anni:", fascia_40_50.shape[0])

#calcola la media della pressione sanguigna diastolica per le diverse fasce d'età

# Fasce di età
fascia_20_29 = df[(df["Age (years)"] >= 20) & (df["Age (years)"] <= 29)]
fascia_30_39 = df[(df["Age (years)"] >= 30) & (df["Age (years)"] <= 39)]
fascia_40_50 = df[(df["Age (years)"] >= 40) & (df["Age (years)"] <= 50)]

# Media della pressione diastolica (mm Hg)
col_pressione = "Diastolic blood pressure (mm Hg)"

print("Pressione media (20-29):", fascia_20_29[col_pressione].mean())
print("Pressione media (30-39):", fascia_30_39[col_pressione].mean())
print("Pressione media (40-50):", fascia_40_50[col_pressione].mean())
```
