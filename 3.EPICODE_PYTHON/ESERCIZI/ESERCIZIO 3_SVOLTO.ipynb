```python
#ES.4 L'azienda Object SPA ha creato una lista. Trasformala in un array NumPY(casting): e rispondi a domande:
#a) quale è stata la vendita massima mensile? e quella minima?
#b) quali sono le vendite mensili maggiori di 4999 oggetti? E quante ne sono?
#c) quali sono le vendite minori di 3000 oggetti?
#d) in media quanti oggetti sono stati venduti al mese?

import numpy as np

lst= [2000,5500,7200,4320,1280,1900,2500,3900,6410,8150,7100,5350]
lst=np.array(lst)
#a) quale è stata la vendita massima mensile? e quella minima?
vendita_max= np.max(lst)  
vendita_min= np.min(lst)
print("RISPOSTA 1: Valore MAX:",vendita_max,"Valore MIN:",vendita_min)
#b) quali sono le vendite mensili maggiori di 4999 oggetti? E quante ne sono?
vendite_filtrate = lst[lst>4999] #creo una variabile con la condizione che cerco nella lista
vendite_mensili = len(vendite_filtrate) # creo una variaile che conta i valori della variabile precedente
print("RISPOSTA 2: Le vendite maggiori di 4999 oggetti sono:",vendite_mensili,"cioè:",vendite_filtrate)
#c) quali sono le vendite minori di 3000 oggetti?
vendite_filtrate_inferiori = lst[lst<3000]
vendite_mensili_inferiori = len(vendite_filtrate_inferiori)
print("RISPOSTA 3: Le vendite minori di 3000 oggetti sono:",vendite_mensili_inferiori,"cioè:",vendite_filtrate_inferiori)
#d) in media quanti oggetti sono stati venduti al mese?
media_oggetti_venduti= np.mean(lst)
print("RISPOSTA 4: La media degli oggetti venduti è",media_oggetti_venduti)
```

```python
#ES.6. L'azienda Object SpA chiede di raddoppiare tutti gli stipendi. Con ciclo for e con maskin
import numpy as np
#ciclo for
stipendi=np.array([100,200,300,400,500,600,700,800,900,1000])
for stipendio in stipendi:
    stipendio= stipendi*2
print("stipendi raddoppiati con ciclo for:",stipendio)

#masking - è una tecnica per filtrare dati utilizzando condizioni logiche.Si basa sulla creazione di maschere booleane (True/False) che
#indicano quali elementi tenere o scartare.

stipendi=np.array([100,200,300,400,500,600,700,800,900,1000])
masking= stipendi >= 100
stipendi[masking]= stipendi[masking]*2
print("stipendi raddoppiati con masking:",stipendi)
```
