```python
#ES.7 - CAMBIARE FORMATI
my_string= "I am studying Python"
print(my_string.upper()) # TUTTI I CARATTERI IN MAIUSCOLO
print(my_string.lower()) # tutti i caratteri in minuscolo
# .strip() - serve per "PULIRE" i caratteri bianchi di una stringa o eliminare gli spazi in eccesso
print(my_string.strip())

#.split() - serve a DIVIDERE una striga in più parti creando una LISTA, di defoult divide una stringa dove trova uno spazio
frase = my_string.split()
print(frase)
```
