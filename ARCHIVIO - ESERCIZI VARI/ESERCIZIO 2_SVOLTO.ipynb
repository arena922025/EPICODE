```python
#ES.10/1 - REALIZZA 2 VERSIONI UNA CON CICLO WHILE E UNA CON CICLO FOR DOVE DEVI:
#CALCOLARE MA NON STAMPARE LE PRIME N POTENZE DI K; OGNIUNA VA MEMORIZZARA, ALLA FINE STAMPA LA LISTA
#RISULTANTE. DA FARE CON VALORI DIVERSI O CON INPUT
K=5
N=9
risultato=[]
potenza=0

while potenza<N:
    calcolo= K**potenza
    risultato.append(calcolo)
    potenza+=1
print("risultato con ciclo while=",risultato)
```

```python
#ES.10/2 - REALIZZA 2 VERSIONI UNA CON CICLO WHILE E UNA CON CICLO FOR DOVE DEVI:
#CALCOLARE MA NON STAMPARE LE PRIME N POTENZE DI K; OGNIUNA VA MEMORIZZARA, ALLA FINE STAMPA LA LISTA
#RISULTANTE. DA FARE CON VALORI DIVERSI O CON INPUT
K=6
N=4
risultato=[]


for calcolo in range(N):
    potenza= K**calcolo
    risultato.append(potenza)
    potenza+=1
print("risultato con ciclo for=",risultato)
```

```python
#ES.12 - ESERCIZIO LISTA DI PAROLE: STAMPA PER OGNI PAROLA QUANTE VOLTE COMPARE LA LETTERA"E", falla apparire sia in maiuscolo che in minuscolo
parole= ["Albergo", "Sedia","Borgo","Petalo","Eremo","Belvedere","Semestre", "Esteta","Sosta", "Ordello", "Abete", "Orologio", "Cesta","Ermellino"]

for parola in parole:
    conto_e = 0
    for lettera in parola:
        if lettera.lower() == "e":
            conto_e += 1
    print(f"La parola '{parola}' contiene {conto_e} lettere 'e'/'E'")
```
